```python
import os
import requests

# Custom headers to satisfy Wikimedia's User-Agent policy
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

def download_image(url, filepath):
    try:
        response = requests.get(url, headers=headers, stream=True, timeout=10)
        response.raise_for_status()
    except Exception as e:
        print(f"Failed to download: {url}\nError: {e}")
        return
    with open(filepath, 'wb') as f:
        for chunk in response.iter_content(chunk_size=1024):
            if chunk:
                f.write(chunk)
    print(f"Downloaded: {filepath}")

def main():
    os.makedirs("images_olimp", exist_ok=True)
    images = {
        "Lenna.png": "https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png",
        "Convolution_arithmetic.png": "https://upload.wikimedia.org/wikipedia/commons/thumb/2/20/Convolution_operation.svg/1200px-Convolution_operation.svg.png",
        "YOLO_object_detection_example.jpg": "https://upload.wikimedia.org/wikipedia/commons/1/15/YOLOv3_results.jpg",
        "MnistExamples.png": "https://upload.wikimedia.org/wikipedia/commons/2/27/MnistExamples.png",
        "Pooling_layer_diagram.png": "https://upload.wikimedia.org/wikipedia/commons/thumb/e/e7/Max_pooling_operation.png/800px-Max_pooling_operation.png",
        "Typical_cnn.png": "https://raw.githubusercontent.com/ageron/handson-ml2/master/images/lenet_architecture.png",
        "Fashion_MNIST_sprite.png": "https://raw.githubusercontent.com/zalandoresearch/fashion-mnist/master/doc/img/fashion-mnist-sprite.png",
        "CIFAR10_img.png": "https://www.cs.toronto.edu/~kriz/cifar-10-sample.png"
    }
    
    for filename, url in images.items():
        filepath = os.path.join("images_olimp", filename)
        download_image(url, filepath)
```

```python
main()
```
